```python
import geopandas as gpd
import json
```

```python
centroids = {}
chicago = gpd.read_file('tracts.geojson')

for row in chicago.iloc:
    centroid = row.geometry.centroid
    centroids[row.namelsad10] = [centroid.x, centroid.y]
```

```python
json.dump(centroids,open('./tract-centroids.json', 'w'),)
```
