```python
import threading
from time import sleep
query = "hospital"

tracts = [key for key, coords in census_tract_centroids.items()]

json.dump(tracts, open('tracts_to_compute.json', 'w'))

def pick_tract_to_compute():
    tracts = json.load(open('tracts_to_compute.json', 'r'))
    tract = tracts[0]
    tracts.remove(tract)
    json.dump(tracts, open('tracts_to_compute.json', 'w'))
    return tract

def compute_places():
    tract_to_compute = pick_tract_to_compute()
    places = utils.places_of_interest.get_places_from_coordinates(census_tract_centroids[tract_to_compute], query, zoom_level=13, amount_of_data=3)
    results[tract_to_compute] = places
    json.dump(results, open(f'../computation_results/{query}.json', 'w'))
    return tract_to_compute

def thread_compute(index):
    print(len(json.load(open('tracts_to_compute.json', 'r'))))
    print(compute_places(), index)
    return

threads = list()

while len(json.load(open('tracts_to_compute.json', 'r'))):
    for index in range(3):
        x = threading.Thread(target=thread_compute, args=(index,))
        sleep(0.2)
        x.start()
```
