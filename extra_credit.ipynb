```python
import pandas as pd
import scipy.stats as stats
import numpy as np

def shift_df(df, amount):
    columns = list(df.columns)
    
    for i in range(amount):
        columns.insert(0,columns.pop())
    return df[columns]
```

## Question 4:

To find whether majority black zip codes were ticketed more often, we decided to use tickets per capita to adjust for population size. Then we find the mean and sample variance for both black and white majority zips, and do a T-test to determine the significance of our findings.

```python
#Question 4:
df = pd.DataFrame(index=majority.keys(),columns=["population", "tickets", "tickets_per_cap", "majority"])


#finding tickets per capita for every zip code
for zip in df.index:
    df['population'][zip] = pop.set_index('Zip')['Estimate!!Total:'][zip]
    df['tickets'][zip] = tickets_per_zip[zip] if zip in tickets_per_zip else 0
    df['tickets_per_cap'][zip] = df['tickets'][zip]/df['poplulation'][zip]
    df['majority'][zip] = majority[zip]
```

```python
#finding mu1 and mu2
mean_tpc_white = df.loc[df['majority'] == 'white']['tickets_per_cap'].mean()
mean_tpc_black = df.loc[df['majority'] == 'black']['tickets_per_cap'].mean()

print(f'Tickets per capita for majority white zip codes = {round(mean_tpc_white, 5)}')
print(f'Tickets per capita for majority black zip codes = {round(mean_tpc_black, 5)}')
```

```python
n1 = len(df.loc[df['majority'] == 'white'].index)
n2 = len(df.loc[df['majority'] == 'black'].index)

var1 = np.var(df.loc[df['majority'] == 'white']['tickets_per_cap'])
var2 = np.var(df.loc[df['majority'] == 'black']['tickets_per_cap'])

v = int((var1/n1+var2/n2)/((var1/n1)**2/(n1-1)+(var2/n2)**2/(n2-1)))

#print(var1, var2, n1, n2)

print(f'Approximate degrees of freedom = {v}')

T0 = (mean_tpc_white-mean_tpc_black)/np.sqrt(var1/n1+var2/n2)

print(f'T0 = {round(T0, 3)}')

for alpha in [1/100, 5/100, 10/100]:
    
    tav = stats.t.ppf(1-alpha, df=v)
    status = "Conclusive" if -tav>T0 else "Inconclusive"
    print(f'{status} for alpha = {alpha*100}%, t={round(tav, 3)}')
```

```python
#Using scipy to confirm our findings and find the p-value
result = stats.ttest_ind(list(df.loc[df['majority'] == 'white']['tickets_per_cap']), list(df.loc[df['majority'] == 'black']['tickets_per_cap']), equal_var=False, alternative='less')

print(f"Using scipy, we get T0={round(result.statistic, 3)}, with p_value= {result.pvalue}")
```
