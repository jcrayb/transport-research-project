```python
import plotly.graph_objects as go
import pandas as pd
import numpy as np
```

```python
results = pd.read_csv('./weighted.csv')

results.head()
```

```python
fig = go.Figure()

red = '#f8766d'
blue = '#619cff'

xmax = results.shortest_path.max()

xw = results.loc[results.majority == 'white'].shortest_path
yw = results.loc[results.majority == 'white'].n_cameras

cw = np.polyfit(xw, yw, 1) 

fig.add_trace(go.Scatter(
    x=xw, 
    y=yw,
    mode='markers',
    marker=dict(
        color=red,
        size=8
    ),
     name='White'
    ))

xb = results.loc[results.majority == 'black'].shortest_path
yb = results.loc[results.majority == 'black'].n_cameras

cb = np.polyfit(xb, yb, 1) 

fig.add_trace(go.Scatter(
    x=xb,
    y=yb,
    mode='markers',
    marker=dict(
        color=blue,
        size=8
    ),
    name='Black'
    ))

fig.add_shape(type='line', x0=0, y0=cb[1], x1=xmax, y1=cb[1]+xmax*cb[0], line=dict(color=blue, width=4))
fig.add_shape(type='line', x0=0, y0=cw[1], x1=xmax, y1=cw[1]+xmax*cw[0], line=dict(color=red, width=4))

fig.update_layout(
        #xaxis=dict(title="Day of Week",
                   #x=0.5),
        yaxis=dict(title="Number of ATLE devices"),
        xaxis=dict(title="Time to reach destination"),
        height=500, 
        width=1200,
        #legend_title_text='Week type',
        margin=dict(l=75, r=10, t=50, b=80),
        legend=dict(
                yanchor="bottom",
                y=.84,
                xanchor="left",
                x=1,
                orientation="v",
                font=dict(
                    #family="Courier",
                    size=20,
                   # color="black"
                ),
            ),
        font=dict(size=20), plot_bgcolor="#ededed",
    )

fig.write_image('./figure/interaction.png', scale=3)
```
