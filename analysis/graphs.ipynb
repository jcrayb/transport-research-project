```python
import plotly.graph_objects as go
import pandas as pd
import numpy as np
import sympy as sp
from sympy.utilities.lambdify import lambdify

res = pd.read_csv('weighted_refactored_income.csv')
res = res.dropna()
res.loc[res.shortest_path == res.shortest_path.max()]
```

```python
def generate_trendline(f, min, max, res=1):
    x = sp.Symbol('x')
    xs = np.linspace(int(min), int(max), int((int(max)-int(min))/res)+1)
    f = lambdify(x, f, 'numpy')

    return (xs, f(xs))
```

```python
red = '#f8766d'
blue = '#619cff'

style_dict = {
    'layout.plot_bgcolor': 'rgba(0, 0, 0, 0)',
    'layout.font.family': 'Times New Roman',
    'layout.font.size': 18,
    'layout.xaxis.linecolor': 'black',
    'layout.xaxis.ticks': 'inside',
    'layout.xaxis.mirror': True,
    'layout.xaxis.showline': True,
    'layout.yaxis.linecolor': 'black',
    'layout.yaxis.ticks': 'inside',
    'layout.yaxis.mirror': True,
    'layout.yaxis.showline': True,
    'layout.autosize': False,
    'layout.legend.bgcolor': 'rgba(0, 0, 0, 0)',
    'layout.legend.xanchor': 'right',
    'layout.legend.x': 1,
    'layout.legend.font.family': 'Times New Roman',
    'layout.legend.font.size': 16,
    'layout.margin': dict(l=75, r=10, t=50, b=80),
}
```

## Commute time per income

```python
x = sp.Symbol('x')

commute_time_per_income = 654.2 - 6.029e-09*x**2

fig = go.Figure()

fig.update(**style_dict)

fig.add_trace(
    go.Scatter(
        x=res.income, y=res.shortest_path,
        mode='markers', 
        marker=dict(
                #color=blue,
                line=dict(width=0)
        )
    )
)

fig.update_layout(
    yaxis=dict(
        title="Average commute time (s)",
        range=[0, 2000]
    ),
    xaxis=dict(
        title="Annual household income (k$)"
    ),
    width=800, 
    height=450,
    showlegend=False
)

trendline = generate_trendline(commute_time_per_income, 5000, 250000)

fig.add_trace(
    go.Scatter(x=trendline[0], y=trendline[1], marker=dict(
                        color=red,
                        line=dict(color='rgb(248, 248, 249)', width=1)
                    ),)
)

fig.write_image('./figure/commuting_time.png', scale=4)
fig.show()
```

```python
fig = go.Figure()

fig.update(**style_dict)

quantiles = [0]+[np.quantile(res.income.to_numpy(), .25*(i+1)) for i in range(4)]

modifier = {
    1: 'st',
    2: 'nd',
    3: 'rd',
    4: 'th'
}

for i in range(1, 5):
    data = res.loc[(res.income <= quantiles[i]) & (res.income > quantiles[i-1])].shortest_path.to_numpy()



    fig.add_trace(
        go.Box(
            y=data,
            name=f"{i}{modifier[i]} quartile",
            boxpoints='outliers', # only outliers
            marker_color='rgb(255,255,255)',
            line_color='#636EFA',
            showlegend=False,
        )
    )

            
fig.update_layout(
    yaxis=dict(
        title="Average commute time (s)",
        range=[0, 1400]
    ),
    xaxis=dict(
        title="Annual household income quartiles"
    ),
    width=800, 
    height=450,
    showlegend=False
)

fig.write_image('./figure/commuting_time_quartiles.png', scale=4)
fig.show()
```
