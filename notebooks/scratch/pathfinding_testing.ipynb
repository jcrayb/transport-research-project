```python
import numpy as np
```

```python
init_nodes = 30

branches = 4
branch_mean = 5
branch_var = 2

mock_init_network = [(i, i+1) for i in range(init_nodes)]

nn = init_nodes

for b in range(branches):
    branch_start = int(np.random.uniform(0, init_nodes-2))

    branch_init_edge = [(branch_start, nn)]
    branch_size = int(np.random.normal(branch_mean, branch_var, 1)[0])

    branch_edges = branch_init_edge + [(nn+i, nn+i+1) for i in range(branch_size)]

    nn += branch_size + 1

    mock_init_network += branch_edges

mock_init_network
```

```python
import networkx as nx

G = nx.DiGraph()

G.add_edges_from(mock_init_network)

nx.draw_networkx(G, pos=nx.spectral_layout(G))
```
