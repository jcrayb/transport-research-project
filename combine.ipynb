```python
import json
import os
```

```python
query = "park"
folder = './computation_results/budget_paths'

res = {}

for file in os.listdir(folder):
    if file[-4:] != "json":
        continue

    if file[:len(query)] != query:
        print(file)
        continue

    data = json.load(open(os.path.join(folder, file), 'r'))

    res = res | data
```

```python
json.dump(res, open(os.path.join('./computation_results/final_results', f'{query}_final_results.json'), 'w'), indent=2)
```
